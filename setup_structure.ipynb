```python
import os

# Two simulation types
sim_types = ["gEEN", "gEIN"]

# Files to create for each simulation type
base_files = [
    "config_{sim}.py",
    "network_{sim}.py",
    "stimulation_{sim}.py",
    "run_{sim}.py",
    "analysis_utils_{sim}.py",
    "utils_{sim}.py"
]

# Template content for each file
template = """# {filename}
# Auto-generated placeholder for the {sim} simulation module.
# You will fill this with the actual simulation code.

def placeholder():
    print("Loaded {filename}")
"""

# Create files
for sim in sim_types:
    for pattern in base_files:
        filename = pattern.format(sim=sim)
        if not os.path.exists(filename):
            with open(filename, "w") as f:
                f.write(template.format(filename=filename, sim=sim))
            print(f"Created {filename}")
        else:
            print(f"Skipped {filename} (already exists)")
```
